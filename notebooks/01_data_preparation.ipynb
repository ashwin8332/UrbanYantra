Ingest Landsat 8 LST, Sentinel-2 indices, WorldCover, ERA5-Land, GHSL, and OSM morphology.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))

from utils import load_config
from data_ingestion import run_ingestion
from features.build_feature_table import build_feature_stack

config = load_config(ROOT / 'configs/pilot_city.yaml')
config['data_mode'] = 'synthetic'  # change to 'gee' after earthengine authenticate
run_ingestion(config)
df, lst_path = build_feature_stack(config, ROOT / config['output_dir'])
df.head()
```
